# Hugging Face Embeddings

## What is Hugging Face?

**Hugging Face** is an open platform whose **Model Hub** hosts thousands of pre-trained models, including many embedding models. Most of them can be downloaded and run **locally for free**, with no API key.

## Why use Hugging Face embeddings?

| | Hugging Face (local) | OpenAI (API) |
|---|---|---|
| **Cost** | Free | Paid per token |
| **API key** | Not needed | Required |
| **Privacy** | Text stays on your machine | Text is sent to OpenAI |
| **Speed** | Depends on your CPU/GPU | Network round-trip |
| **Setup** | Model downloads on first use | None |

## The `sentence-transformers` library

Most Hugging Face embedding models are used through `sentence-transformers`. It handles the tokenize → encode → pool steps from the previous notebook in a single call:

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
vectors = model.encode(["How do I reset my password?", "I forgot my login"])
```

The model is downloaded from the Hub once and cached locally in `~/.cache/huggingface/`.

## Popular models

| Model | Dimensions | Notes |
|---|---|---|
| `sentence-transformers/all-MiniLM-L6-v2` | 384 | Small and fast, a good default for learning |
| `sentence-transformers/all-mpnet-base-v2` | 768 | Higher quality, slower |
| `BAAI/bge-small-en-v1.5` | 384 | Strong retrieval quality for its size |
| `BAAI/bge-base-en-v1.5` | 768 | Larger BGE model, better accuracy |

> To compare models, see the [MTEB leaderboard](https://huggingface.co/spaces/mteb/leaderboard), which ranks embedding models on retrieval and other tasks.

## Using Hugging Face embeddings in LangChain

LangChain's `HuggingFaceEmbeddings` wraps `sentence-transformers` and exposes the same interface as every other LangChain embedding class:

- `embed_query(text)` returns **one** vector (for the user's question)
- `embed_documents(texts)` returns **a list** of vectors (for document chunks)

Because the interface is shared, you can later swap in `OpenAIEmbeddings` without changing the rest of your RAG pipeline.

> `HuggingFaceEmbeddings` lives in the `langchain-huggingface` package, which isn't in this project yet. Install it with `uv add langchain-huggingface`.

````python
from langchain_huggingface import HuggingFaceEmbeddings

## Initilize a simple Embedding model(No API Key needed)

embeddings = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
embeddings
````

````python
## Create embeddings

text = "Hello, This is test notebook for embeddings!."
embedding = embeddings.embed_query(text)

print(f"Text : {text}")
print(f"Embeddings length : {len(embedding)}")
print(embedding)
````

````python
sentences = [
    "The cat is sleeping on the sofa.",
    "A kitten is napping on the couch.",
    "How do I reset my password?",
    "I forgot my login credentials.",
    "The stock market dropped sharply today.",
]

vectors = embeddings.embed_documents(sentences)

print(f"Number of sentences : {len(sentences)}")
print(f"Number of vectors   : {len(vectors)}")
print(f"Vector length       : {len(vectors[0])}")
print(f" Vectors : {vectors}")
````

````python
import numpy as np

def cosine_similarity(a, b):
    a, b = np.array(a), np.array(b)
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
````

````python
## Multilingual model: same meaning in different languages gets similar vectors

multilingual_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

english = "The weather is beautiful today."
other_sentences = {
    "Spanish":   "El clima está hermoso hoy.",
    "French":    "Il fait très beau aujourd'hui.",
    "German":    "Das Wetter ist heute wunderschön.",
    "Tamil":     "இன்று வானிலை மிகவும் அழகாக இருக்கிறது.",
    "Unrelated": "I need to fix my car's engine.",
}

english_vector = multilingual_model.embed_query(english)

print(f"English: {english}\n")
for language, sentence in other_sentences.items():
    score = cosine_similarity(english_vector, multilingual_model.embed_query(sentence))
    print(f"{language:<10} {score:.3f}  {sentence}")
````

````python
## all-MiniLM-L6-v2 was trained on English, so it struggles to match translations

spanish = other_sentences["Spanish"]
english_only_score = cosine_similarity(embeddings.embed_query(english), embeddings.embed_query(spanish))
multilingual_score = cosine_similarity(english_vector, multilingual_model.embed_query(spanish))

print(f"English vs Spanish with all-MiniLM-L6-v2 : {english_only_score:.3f}")
print(f"English vs Spanish with multilingual     : {multilingual_score:.3f}")
````

````python
## Q&A model: rank passages by how well they answer a question

qa_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/multi-qa-MiniLM-L6-cos-v1"
)

question = "How can I get my money back?"

passages = [
    "Refunds are issued within 5-7 business days after we receive the returned item.",
    "Our store opens at 9 AM and closes at 8 PM on weekdays.",
    "To reset your password, click 'Forgot password' on the login page.",
    "Shipping is free for orders over $50.",
    "You can return any product within 30 days for a full refund.",
]

question_vector = qa_model.embed_query(question)
passage_vectors = qa_model.embed_documents(passages)

scores = [cosine_similarity(question_vector, pv) for pv in passage_vectors]
ranked = sorted(zip(scores, passages), reverse=True)

print(f"Question: {question}\n")
for rank, (score, passage) in enumerate(ranked, start=1):
    print(f"{rank}. {score:.3f}  {passage}")
````
